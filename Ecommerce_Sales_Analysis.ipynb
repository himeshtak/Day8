### Day 8 — Pandas Data Analysis Notebook

This notebook analyzes an e-commerce sales dataset using **Pandas**. It covers:

1. Loading & inspecting the dataset
2. Data selection & filtering
3. Sorting operations
4. `groupby()` and aggregation functions
5. Category-wise, city-wise, product-wise, and payment-method-wise analysis

7. Top performers and specific-order lookups
8. Final observations

**Dataset:** `Day8_Ecommerce_Sales_Dataset.csv`


## 1. Setup & Load Data

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)

# Load the dataset
df = pd.read_csv('Day8_Ecommerce_Sales_Dataset.csv')

print("Shape of dataset:", df.shape)
df.head()
```

```python
# Basic info about the dataset
df.info()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Convert Order_Date to datetime for proper handling
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.dtypes
```

```python
# Statistical summary of numeric columns
df.describe()
```

## 2. Data Selection
Selecting specific columns and rows using label-based (`.loc`) and position-based (`.iloc`) indexing.

```python
# Select specific columns
df[['Order_ID', 'Product', 'Category', 'City', 'Total_Sales']].head()
```

```python
# Select rows using iloc (first 5 rows, first 6 columns)
df.iloc[0:5, 0:6]
```

```python
# Select rows using loc with a condition, showing specific columns
df.loc[df['Category'] == 'Electronics', ['Product', 'City', 'Total_Sales', 'Payment_Method']].head()
```

## 3. Filtering Data

```python
# Orders where Total_Sales is greater than 5000
high_value_orders = df[df['Total_Sales'] > 5000]
print(f"Number of high-value orders (> 5000): {len(high_value_orders)}")
high_value_orders[['Order_ID', 'Product', 'City', 'Total_Sales']].head()
```

```python
# Orders paid via UPI in the South region
upi_south = df[(df['Payment_Method'] == 'UPI') & (df['Region'] == 'South')]
print(f"UPI orders from South region: {len(upi_south)}")
upi_south[['Order_ID', 'Customer', 'City', 'Product', 'Total_Sales']]
```

```python
# Orders with a discount applied (Discount_Percent > 0) and rating >= 4.5
good_discounted_orders = df[(df['Discount_Percent'] > 0) & (df['Rating'] >= 4.5)]
print(f"Discounted orders with rating >= 4.5: {len(good_discounted_orders)}")
good_discounted_orders[['Order_ID', 'Product', 'Discount_Percent', 'Rating', 'Total_Sales']]
```

```python
# Orders for the 'Electronics' category with Quantity greater than 2
electronics_bulk = df[(df['Category'] == 'Electronics') & (df['Quantity'] > 2)]
electronics_bulk[['Order_ID', 'Product', 'City', 'Quantity', 'Total_Sales']]
```

## 4. Sorting Data

```python
# Top 10 orders by Total_Sales (descending)
top10_sales = df.sort_values(by='Total_Sales', ascending=False).head(10)
top10_sales[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python
# Lowest 5 orders by Total_Sales (ascending)
lowest5_sales = df.sort_values(by='Total_Sales', ascending=True).head(5)
lowest5_sales[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python
# Sort by Category (asc) then Total_Sales (desc) - useful for viewing top orders within each category
df.sort_values(by=['Category', 'Total_Sales'], ascending=[True, False]).head(10)[
    ['Category', 'Product', 'City', 'Total_Sales']]
```

```python
# Sort by Rating to see best and worst rated orders
best_rated = df.sort_values(by='Rating', ascending=False).head(5)
best_rated[['Order_ID', 'Product', 'Rating', 'Total_Sales']]
```

## 5. Overall Key Statistics
Total sales, average sales, highest/lowest sales, total quantity sold, and number of orders — for the entire dataset.

```python
total_sales = df['Total_Sales'].sum()
average_sales = df['Total_Sales'].mean()
highest_sale = df['Total_Sales'].max()
lowest_sale = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
number_of_orders = df['Order_ID'].nunique()

print(f"Total Sales          : ₹{total_sales:,.2f}")
print(f"Average Sales/Order   : ₹{average_sales:,.2f}")
print(f"Highest Single Sale   : ₹{highest_sale:,.2f}")
print(f"Lowest Single Sale    : ₹{lowest_sale:,.2f}")
print(f"Total Quantity Sold   : {total_quantity} units")
print(f"Number of Orders      : {number_of_orders}")
```

```python
# The order with the highest and lowest sale amount
highest_order = df.loc[df['Total_Sales'].idxmax()]
lowest_order = df.loc[df['Total_Sales'].idxmin()]

print("Highest sale order:")
print(highest_order[['Order_ID', 'Product', 'City', 'Total_Sales']])
print("\nLowest sale order:")
print(lowest_order[['Order_ID', 'Product', 'City', 'Total_Sales']])
```

## 6. Category-wise Analysis
Using `groupby()` and aggregation functions (`sum`, `mean`, `max`, `min`, `count`) to analyze sales by product category.

```python
category_analysis = df.groupby('Category').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Highest_Sale=('Total_Sales', 'max'),
    Lowest_Sale=('Total_Sales', 'min'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

category_analysis
```

```python
# Best performing category by total sales
top_category = category_analysis['Total_Sales'].idxmax()
print(f"🏆 Top-performing category: {top_category}")
print(f"   Total Sales: ₹{category_analysis.loc[top_category, 'Total_Sales']:,.2f}")
```

## 7. City-wise Analysis

```python
city_analysis = df.groupby('City').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

city_analysis
```

```python
# Top 5 cities by total sales
top5_cities = city_analysis.head(5)
print("🏙️ Top 5 cities by total sales:")
top5_cities
```

```python
# Region-wise sales (Region column also available)
region_analysis = df.groupby('Region').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

region_analysis
```

## 8. Product-wise Analysis

```python
product_analysis = df.groupby('Product').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

product_analysis
```

```python
# Top 5 best-selling products by total sales
top5_products = product_analysis.head(5)
print("🥇 Top 5 products by total sales:")
top5_products
```

```python
# Product with the highest quantity sold
top_qty_product = product_analysis['Total_Quantity'].idxmax()
print(f"📦 Product with highest quantity sold: {top_qty_product} "
      f"({product_analysis.loc[top_qty_product, 'Total_Quantity']} units)")
```

## 9. Payment Method Analysis

```python
payment_analysis = df.groupby('Payment_Method').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'count')
).round(2).sort_values(by='Total_Sales', ascending=False)

payment_analysis
```

```python
# Most popular payment method by number of orders
most_popular_payment = payment_analysis['Number_of_Orders'].idxmax()
print(f"💳 Most used payment method: {most_popular_payment} "
      f"({payment_analysis.loc[most_popular_payment, 'Number_of_Orders']} orders)")
```

## 10. Multi-level Grouping
Combining `groupby()` on multiple columns for deeper insights.

```python
# Category-wise sales broken down by Payment Method
category_payment = df.groupby(['Category', 'Payment_Method'])['Total_Sales'].sum().round(2)
category_payment_df = category_payment.reset_index().sort_values(
    by=['Category', 'Total_Sales'], ascending=[True, False])
category_payment_df
```

```python
# City-wise sales broken down by Category (pivot table view)
city_category_pivot = pd.pivot_table(
    df, values='Total_Sales', index='City', columns='Category',
    aggfunc='sum', fill_value=0
).round(2)
city_category_pivot
```

## 11. Identifying Specific Orders & Top Performers

```python
# Specific order lookup - e.g., Order_ID = 1013
specific_order = df[df['Order_ID'] == 1013]
specific_order
```

```python
# Top 3 products within EACH category (by total sales)
top3_per_category = (
    df.groupby(['Category', 'Product'])['Total_Sales']
    .sum()
    .reset_index()
    .sort_values(['Category', 'Total_Sales'], ascending=[True, False])
    .groupby('Category')
    .head(3)
)
top3_per_category
```

```python
# Orders that are both high value (>7000) AND highly rated (>=4.5)
premium_orders = df[(df['Total_Sales'] > 7000) & (df['Rating'] >= 4.5)].sort_values(
    by='Total_Sales', ascending=False)
print(f"Premium orders found: {len(premium_orders)}")
premium_orders[['Order_ID', 'Product', 'City', 'Total_Sales', 'Rating']]
```

## 12. Summary Table of Overall Key Metrics

```python
summary = pd.DataFrame({
    'Metric': ['Total Sales', 'Average Sales per Order', 'Highest Sale', 'Lowest Sale',
               'Total Quantity Sold', 'Number of Orders', 'Top Category', 'Top City',
               'Top Product', 'Most Used Payment Method'],
    'Value': [
        f"₹{total_sales:,.2f}",
        f"₹{average_sales:,.2f}",
        f"₹{highest_sale:,.2f}",
        f"₹{lowest_sale:,.2f}",
        f"{total_quantity} units",
        f"{number_of_orders}",
        top_category,
        top5_cities.index[0],
        top5_products.index[0],
        most_popular_payment
    ]
})
summary
```

## 13. Observations

Based on the analysis above:

1. **Category performance:** One category clearly generates the highest total sales among all categories (see Section 6 output), indicating strong customer demand relative to other segments — though this should be weighed against the number of orders and average order value per category.
2. **Geographic concentration:** Sales are not evenly distributed across cities — a small number of cities account for a disproportionately large share of total revenue, suggesting these are priority markets for marketing and inventory focus.
3. **Product-level insight:** A handful of products dominate total sales, and the product with the highest total sales isn't always the one with the highest quantity sold — this points to differences in unit price and order size (a good pricing/bundling insight).
4. **Payment behavior:** One payment method is used considerably more than others, which could reflect customer trust/convenience and may guide which payment options to promote or optimize for a smoother checkout experience.
5. **Discount and satisfaction link:** Orders combining a discount with a high customer rating (≥4.5) suggest that reasonable discounting doesn't hurt satisfaction — and may even support it — useful for future promotional strategy.
6. **Order value spread:** There is a wide gap between the highest and lowest single-order sale values, showing the dataset contains both small individual purchases and large bulk/high-ticket orders, which is typical of a mixed-category e-commerce store.
